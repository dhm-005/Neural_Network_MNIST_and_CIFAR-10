# Phân tích dữ liệu

```python
import numpy as np
import struct

def load_images(path):
    with open(path, "rb") as f:
        magic, num_images, rows, cols = struct.unpack(">IIII", f.read(16))

        data = np.frombuffer(
            f.read(),
            dtype=np.uint8
        )

    images = data.reshape(num_images, rows, cols)

    return images
```

```python
def load_labels(path):
    with open(path, "rb") as f:
        magic, num_labels = struct.unpack(">II", f.read(8))

        labels = np.frombuffer(
            f.read(),
            dtype=np.uint8
        )

    return labels
```

```python
X_train = load_images("MNIST/train-images.idx3-ubyte")
y_train = load_labels("MNIST/train-labels.idx1-ubyte")

X_test = load_images("MNIST/t10k-images.idx3-ubyte")
y_test = load_labels("MNIST/t10k-labels.idx1-ubyte")

print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

## CNN PyTorch

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Chuyển về tensor
X_train_t = torch.tensor(X_train, dtype=torch.float32)
X_test_t = torch.tensor(X_test, dtype=torch.float32)

X_train_t = X_train_t.permute(0, 3, 1, 2)
X_test_t = X_test_t.permute(0, 3, 1, 2)

y_train_t = torch.tensor(y_train, dtype=torch.long)
y_test_t = torch.tensor(y_test, dtype=torch.long)

print(X_train_t.shape, X_test_t.shape)
print(y_train_t.shape, y_test_t.shape)
```

```python
class CNN(nn.Module):

    def __init__(self):
        super().__init__()

        # Conv 1
        self.conv1 = nn.Conv2d(1, 8, 3)

        # Pooling
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Conv 2
        self.conv2 = nn.Conv2d(8, 8, 3)

        # Fully Connected
        self.fc1 = nn.Linear(8 * 5 * 5, 16)

        self.fc2 = nn.Linear(16, 10)

    def forward(self, x):

        # Conv 1 → ReLU → Pool
        x = self.conv1(x)
        x = torch.relu(x)
        x = self.pool(x)

        # Conv 2 → ReLU → Pool
        x = self.conv2(x)
        x = torch.relu(x)
        x = self.pool(x)

        # Flatten
        x = x.view(x.size(0),-1)

        # Dense
        x = self.fc1(x)
        x = torch.relu(x)

        # Output
        x = self.fc2(x)

        return x
```

```python
model = CNN()

criterion = nn.CrossEntropyLoss()

optimizer = optim.SGD(
    model.parameters(),
    lr=0.01
)
```

```python
epochs = 10
batch_size = 32

losses = []

for epoch in range(epochs):

    permutation = torch.randperm(
        X_train_t.size(0)
    )

    total_loss = 0
    num_batches = 0

    for i in range(0, X_train_t.size(0), batch_size):

        idx = permutation[i:i + batch_size]

        X_batch = X_train_t[idx]
        y_batch = y_train_t[idx]

        optimizer.zero_grad()

        y_hat = model(X_batch)

        loss = criterion(
            y_hat,
            y_batch
        )

        loss.backward()

        optimizer.step()

        total_loss += loss.item()
        num_batches += 1

    avg_loss = total_loss / num_batches

    losses.append(avg_loss)

    print(
        f"Epoch {epoch + 1}/{epochs} "
        f"- Loss: {avg_loss:.4f}"
    )
```

```python
drawLvsE(losses)
```

## CNN TensorFlow

```python
nf, mf, c = X_train.shape[1:]
```

```python
import tensorflow as tf

cnn_tf = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(nf, mf, c)),
    tf.keras.layers.Conv2D(
        filters=8,
        kernel_size=3,
        activation="relu"
    ),
    tf.keras.layers.MaxPooling2D(c),
    tf.keras.layers.Conv2D(
        filters=8,
        kernel_size=3,
        activation="relu"
    ),
    tf.keras.layers.MaxPooling2D(c),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

cnn_tf.summary()
```

```python
cnn_tf.compile(
    optimizer=tf.keras.optimizers.SGD(
        learning_rate=0.01
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history_cnn = cnn_tf.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    verbose=1
)
```

```python
drawLvsE(history_cnn.history["loss"])
```

## RNN TensorFlow

```python
rnn = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(nf, mf)),
    tf.keras.layers.SimpleRNN(8),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
rnn.summary()
rnn.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history_rnn = rnn.fit(
    X_train, y_train,
    epochs=10,
    batch_size=32,
    verbose=1
)
```

```python
drawLvsE(history_rnn.history["loss"])
```

## LSTM TensorFlow

```python
lstm = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(nf, mf)),
    tf.keras.layers.LSTM(8),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

lstm.compile(
    optimizer="sgd",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history_lstm = lstm.fit(
    X_train, y_train,
    epochs=10,
    batch_size=32,
    verbose=1
)
```

```python
drawLvsE(history_lstm.history["loss"])
```

# Kết quả

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix
)

def evaluation(name, y_test, y_prob):
    y_pred = np.argmax(y_prob, axis=1)

    acc = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, average='weighted', zero_division=0)
    recall = recall_score(
        y_test, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(
        y_test, y_pred, average='weighted', zero_division=0)
    cm = confusion_matrix(y_test, y_pred)

    return {
        'Model': name,
        'Accuracy': acc, 'Precision': precision,
        'Recall': recall, 'F1-score': f1,
        'Confusion Matrix': cm
    }
```

```python
res = []
# CNN from scratch
res.append(cnn_res)

# CNN PyTorch
with torch.no_grad():
    y_prob = model(X_test_t)
y_prob = y_prob.numpy()
res.append(evaluation("CNN PyTorch", y_test, y_prob))

#CNN TensorFlow
y_prob = cnn_tf.predict(X_test,verbose=0)
res.append(evaluation("CNN TensorFlow", y_test, y_prob))

#RNN TensorFlow
y_prob = rnn.predict(X_test, verbose=0)
res.append(evaluation("RNN TensorFlow", y_test, y_prob))

#LSTM TensorFlow
y_prob = lstm.predict(X_test, verbose=0)
res.append(evaluation("LSTM TensorFlow", y_test, y_prob))
```

```python
import pandas as pd

res_df = pd.DataFrame(res).set_index('Model')

metric_df = res_df.drop(columns = ['Confusion Matrix', 'Losses'])

metric_df.sort_values(
    by='Accuracy',
    ascending=False
)
```
